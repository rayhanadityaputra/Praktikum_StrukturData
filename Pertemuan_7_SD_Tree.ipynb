**1. Membuat Binary Tree Secara Manual**

```python
# @title Binary Tree
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right =  None

# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right =  Node(5)

# Fungsi inorder traversal
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=" ")
    inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

**2. Membuat Binary Search Tree**

```python
class Node:
  def __init__(self,data):
    self.data = data
    self.left = None
    self.right = None

class BinarySearchTree:
  def __init__(self):
    self.root = None
  def insert(self, root, data):
    if root is None:
      return Node(data)
    if data < root.data:
      root.left = self.insert(root.left, data)
    else:
      root.right = self.insert(root.right, data)
    return root

bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
  root = bst.insert(root, value)
```

**3. Traversal pada BST**

```python
def inorder(node):
 if node:
  inorder(node.left)
  print(node.data, end=' ')
  inorder(node.right)

def preorder(node):
 if node:
  print(node.data, end=' ')
  preorder(node.left)
  preorder(node.right)

def postorder(node):
 if node:
  postorder(node.left)
  postorder(node.right)
  print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

**4. Pencarian Nilai dalam BST**

```python
def search(node, key):
 if node is None or node.data == key:
  return node
 if key < node.data:
  return search(node.left, key)
 return search(node.right, key)

# Uji pencarian
key = 60
result = search(root, key)
if result:
 print(f"{key} ditemukan dalam tree.")
else:
 print(f"{key} tidak ditemukan.")

key = 25
result = search(root, key)
if result:
 print(f"{key} ditemukan dalam tree.")
else:
 print(f"{key} tidak ditemukan.")
```

```python
# @title Tugas 2: Program Binary Search Tree (BST)

# Program Binary Search Tree (BST)
# Data berdasarkan identitas pribadi

# =================================
# Membuat class Node
# =================================
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# =================================
# Membuat class BST
# =================================
class BinarySearchTree:
    def __init__(self):
        self.root = None

    # Fungsi insert data ke BST
    def insert(self, root, data):

        # Jika node kosong
        if root is None:
            return Node(data)

        # Jika data lebih kecil masuk kiri
        if data < root.data:
            root.left = self.insert(root.left, data)

        # Jika data lebih besar masuk kanan
        else:
            root.right = self.insert(root.right, data)

        return root


# =================================
# Fungsi Traversal
# =================================

# Inorder Traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)


# Preorder Traversal
def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)


# Postorder Traversal
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')


# =================================
# Fungsi Pencarian Data
# =================================
def search(node, key):

    # Jika data ditemukan atau node kosong
    if node is None or node.data == key:
        return node

    # Cari ke subtree kiri
    if key < node.data:
        return search(node.left, key)

    # Cari ke subtree kanan
    return search(node.right, key)


# =================================
# Data BST
# =================================

# 1. Tiga digit terakhir NPM = 005 -> 5
# 2. Tanggal lahir = 02 dan 10
# 3. ASCII huruf pertama dan kedua nama depan
#    R = 82
#    a = 97
# 4. Dua angka favorit
#    25 dan 99

data_list = [5, 2, 10, 82, 97, 25, 99]


# =================================
# Membentuk BST
# =================================
bst = BinarySearchTree()
root = None

for data in data_list:
    root = bst.insert(root, data)


# =================================
# Menampilkan Traversal
# =================================
print("Inorder Traversal:")
inorder(root)

print("\n\nPreorder Traversal:")
preorder(root)

print("\n\nPostorder Traversal:")
postorder(root)


# =================================
# Pencarian Data
# =================================

# Cari dua digit terakhir NPM = 05
key = 5

result = search(root, key)

if result:
    print(f"\n\nData {key} ditemukan dalam BST.")
else:
    print(f"\n\nData {key} tidak ditemukan.")


# Cari angka yang tidak ada dalam list
key = 50

result = search(root, key)

if result:
    print(f"Data {key} ditemukan dalam BST.")
else:
    print(f"Data {key} tidak ditemukan.")
```
