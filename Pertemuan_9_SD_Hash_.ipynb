**1. Implementasi Hash Table Sederhana**

```python
class SimpleHashTable:
  def __init__ (self, size):
    self.size = size
    self.table = [None] * size

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    self.table[index] = value

  def search(self, key) :
    index = self.hash_function(key)
    return self.table[index]

# Contoh penggunaan
ht = SimpleHashTable(10)
ht.insert ("apel", 100)
ht.insert("pisang", 200)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'pisang':", ht.search("pisang"))
```

```python
class LinearProbingHashTable:
  def __init__ (self, size):
    self.size = size
    self.table = [None] * size

  def hash_function (self, key):
    return hash(key) % self.size

  def insert (self, key, value):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index] [0] == key:
        break
      index = (index + 1) % self.size
      if index == original_index:
        raise Exception("Hash table penuh")

    self.table[index] = (key, value)

  def search(self, key):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index] [0] == key:
        return self.table[index] [1]
      index = (index + 1) % self.size
      if index == original_index:
        break
    return None

# Contoh penggunaan
ht = LinearProbingHashTable (5)
ht.insert ("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search("melon"))
```

**3. Penanganan Collision dengan Chaining**

```python
class ChainingHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [[] for _ in range (size)]

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    for i, (k, v) in enumerate (self.table[index]):
      if k == key:
        self.table[index] [i] = (key, value)
        return
    self.table[index].append((key, value))

  def search(self, key):
    index = self.hash_function(key)
    for k, v in self.table[index]:
      if k == key:
        return v
    return None

# Contoh penggunaan
ht = ChainingHashTable(5)
ht.insert ("apel", 100)
ht.insert ("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search ("melon"))
```

**Latihan / Tugas**

```python
# @title 1. Eksperimen Sederhana Linear Probing

class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                break

            index = (index + 1) % self.size

            if index == original_index:
                raise Exception("Hash table penuh")

        self.table[index] = (key, value)

    def search(self, key):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                return self.table[index][1]

            index = (index + 1) % self.size

            if index == original_index:
                break

        return None


# Membuat objek hash table ukuran 5
ht = LinearProbingHashTable(5)

# Menambahkan data
ht.insert("A", 10)
ht.insert("B", 20)
ht.insert("C", 30)
ht.insert("D", 40)
ht.insert("E", 50)

# Menampilkan isi tabel
print("Isi hash table:")
print(ht.table)
```

**Penjelasan :**

Jika dua key memiliki hasil hash yang sama, maka terjadi collision.
Pada metode Linear Probing, collision ditangani dengan mencari slot kosong berikutnya secara berurutan dari index awal. Proses pencarian dilakukan dengan menambah index satu per satu hingga ditemukan posisi kosong untuk menyimpan data baru.

```python
# @title 2. Percobaan Dasar Chaining

class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)

        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return

        self.table[index].append((key, value))

    def search(self, key):
        index = self.hash_function(key)

        for k, v in self.table[index]:
            if k == key:
                return v

        return None


# Membuat objek hash table ukuran 3
ht = ChainingHashTable(3)

# Menambahkan data
ht.insert("apel", 100)
ht.insert("melon", 200)
ht.insert("lemon", 300)
ht.insert("pisang", 400)

# Menampilkan isi tabel
print("Isi hash table:")
for i, data in enumerate(ht.table):
    print(f"Index {i}: {data}")
```

**Penjelasan :**

Pada metode Chaining, collision ditangani dengan menyimpan beberapa data dalam satu slot menggunakan list.
Jika dua atau lebih key memiliki index hash yang sama, maka data tersebut akan ditambahkan ke list pada index tersebut. Dengan demikian, satu slot dapat menyimpan banyak pasangan key-value tanpa menimpa data lain
