**3. Traversal Graf menggunakan DFS**

```python
def dfs (graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph [start]:
      dfs(graph, neighbor, visited)

# Contoh pemanggilan
dfs(graph, 'A')
```

**LATIHAN / TUGAS**

```python
from collections import deque

# =====================================================
# 1. MEMBUAT GRAF TAK BERARAH
# =====================================================

# Graf memiliki:
# - 6 simpul : A, B, C, D, E, F
# - 8 sisi

graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['A', 'B', 'D', 'F'],
    'D': ['A', 'C', 'E'],
    'E': ['B', 'D', 'F'],
    'F': ['C', 'E']
}

# Fungsi untuk menampilkan graf
def print_graph(graf):
    print("Adjacency List Graf:")

    for node in graf:
        print(f"{node} -> {graf[node]}")


# =====================================================
# 2. IMPLEMENTASI BFS DAN DFS
# =====================================================

# -------------------------
# BFS (Breadth First Search)
# -------------------------
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    result = []

    while queue:
        vertex = queue.popleft()

        if vertex not in visited:
            visited.add(vertex)
            result.append(vertex)

            queue.extend(
                [neighbor for neighbor in graph[vertex]
                 if neighbor not in visited]
            )

    return result


# -------------------------
# DFS (Depth First Search)
# -------------------------
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)

        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)


# =====================================================
# 3. MODIFIKASI BFS
# =====================================================

# BFS sudah dimodifikasi agar:
# - Tidak hanya mencetak
# - Tetapi mengembalikan list urutan kunjungan


# =====================================================
# 4. FUNGSI FIND_PATH MENGGUNAKAN DFS
# =====================================================

def find_path(graph, start, end, path=None):

    if path is None:
        path = []

    # Menambahkan simpul ke jalur
    path = path + [start]

    # Jika simpul tujuan ditemukan
    if start == end:
        return path

    # Menelusuri tetangga
    for neighbor in graph[start]:

        if neighbor not in path:

            new_path = find_path(
                graph,
                neighbor,
                end,
                path
            )

            if new_path:
                return new_path

    return None


# =====================================================
# 5. FUNGSI IS_CONNECTED
# =====================================================

def is_connected(graph):

    visited = set()

    # DFS untuk mengecek keterhubungan
    def dfs_connected(node):

        visited.add(node)

        for neighbor in graph[node]:

            if neighbor not in visited:
                dfs_connected(neighbor)

    # Mengambil simpul pertama
    start_node = next(iter(graph))

    # Menjalankan DFS
    dfs_connected(start_node)

    # Membandingkan jumlah simpul yang dikunjungi
    return len(visited) == len(graph)


# =====================================================
# OUTPUT PROGRAM
# =====================================================

# -------------------------
# 1. Menampilkan Graf
# -------------------------
print("1. Menampilkan Graf")
print_graph(graph)

# -------------------------
# 2. Menjalankan BFS
# -------------------------
print("\n2. Hasil BFS dari simpul A")

hasil_bfs = bfs(graph, 'A')

print(hasil_bfs)

# -------------------------
# 3. Menjalankan DFS
# -------------------------
print("\n3. Hasil DFS dari simpul A")

dfs(graph, 'A')

# -------------------------
# 4. Mencari Jalur
# -------------------------
print("\n\n4. Mencari Jalur dari A ke F")

jalur = find_path(graph, 'A', 'F')

print(jalur)

# -------------------------
# 5. Mengecek Keterhubungan
# -------------------------
print("\n5. Mengecek Apakah Graf Terhubung")

if is_connected(graph):
    print("Graf terhubung")
else:
    print("Graf tidak terhubung")
```
